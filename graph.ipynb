```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("src/test/resources/graph.csv")
df = df.replace([np.inf, -np.inf], np.nan).dropna()

fig, ax = plt.subplots(figsize=(12, 6))
ax.scatter(df['x'], df['value'], s=10, color='red', marker='o')

ax.set_xlim(-20, 20)  
ax.set_ylim(-20, 20) 

ax.grid(True, linestyle='--', linewidth=0.5)
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)

ax.set_xlabel("x")
ax.set_ylabel("value")
ax.set_title("График с ограничением осей")
plt.tight_layout()
plt.show()
```
